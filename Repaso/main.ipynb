```python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
```

```python
""" Probando modelos """
#LLM = Large Language Model
#El LLM se basa en el modelo de nuestra IA, puede ser GPT 4.0 Mini o Mistral Nemo de Ollama
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0.5, openai_api_key=api_key)
# llm = ChatOllama(model='mistral-nemo', temperature=0.5)
#Aquí ejecutamos el modelo con un prompt para ver que nos contesta
#Se utiliza invoke que recibe un String, después mostramos el resultado con content
response = llm.invoke('Hola')
response.content
```

```python
""" Chain básica """
#Una Chain es el flujo que debe tener el chatbot
#Esta cadena de trabajo debe tener una plantilla o template que sirve para dar estructura a las consultas
from langchain.prompts import PromptTemplate
```

```python
#Creamos una variable con un template que recibe un tema de conversación
template = 'Dame un resumen breve sobre {topic}'
#El PromptTemplare recibe un tema y la estructura de la cadena o consulta
prompt = PromptTemplate(input_variables=['topic'], template=template)
```

```python
# Crear Chain
#La Chain recibe el prompt template y también el modelo de LLM
chain = prompt | llm
```

```python
# Ejecutar Chain
# Para ejecutar la chain usamos invoke, y le pasamos nuestro tema de conversación, que como ya indicamos recibe una variable
response = chain.invoke(
    input={'topic': 'Videojuego Halo'}
)

#El invoke solo es para ejecutar y guardar la respuesta del Chat
#Se muestra la respuesta con el content
response.content
```

```python
""" SequentialChain """
# Chain 1: Generar título
title_prompt = PromptTemplate(
    input_variables=['topic'],
    template='Genera un título atractivo sobre {topic}.'
)
title_chain = title_prompt | llm

# Chain 2: Crear resumen
summary_prompt = PromptTemplate(
    input_variables=['title'],
    template='Escribe un resumen breve para el título: {title}.'
)
summary_chain = summary_prompt | llm
```

```python
# SequentialChain
sequential_chain = title_chain | summary_chain

# Ejecutar Chain
response = sequential_chain.invoke(
    input={'topic': 'Vehículos eléctricos'}
)
response.content
```
